```python
import sys, json; sys.path.insert(0, '../scripts')
from clinker_phases import hydrate
for name, comp in {'A (55·18·10·8)': (55, 18, 10, 8), 'B (30·45·4·12)': (30, 45, 4, 12), 'W01 (55·20·10·8)': (55, 20, 10, 8)}.items():
    r = hydrate(*comp); print(name, {k: round(v, 1) for k, v in r['products_g'].items()}, {k: round(v, 1) for k, v in r['consumed_g'].items()}, '결합수/클링커', round(r['ratios']['결합수/클링커'], 3))
```

```python
MY = (55, 18, 10, 8)   # 내 조성 wt% — 바꾼다
r = hydrate(*MY); json.dump(r, open('../out/hydration_products.json', 'w'), indent=1, ensure_ascii=False); print('saved out/hydration_products.json')
```

```python
from csh_layer_model import model
for cs in (0.83, 1.0, 1.5, 1.7):
    m = model(cs, water=2.0); print(cs, '→ Ca/Si', m['Ca/Si'], '· 가교 제거', m['bridge_removed'], '· 사슬 길이', m['mean_chain_length_tetra'], '· 층간 Ca', m['Ca_interlayer'], '· 층간 물', m['interlayer_H2O'])
json.dump({'1.0': model(1.0), '1.7': model(1.7)}, open('../out/csh_model.json', 'w'), indent=1, ensure_ascii=False); print('saved out/csh_model.json')
```
