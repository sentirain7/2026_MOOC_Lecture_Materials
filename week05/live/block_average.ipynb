# W05 · 라이브 블록 평균 (덱 15장과 같은 계산)
W03 온도 시계열 1,001점 · 폐기 구간 t_eq 와 블록 크기 b 를 바꿔 SE · N_eff 확인

```python
import numpy as np, matplotlib.pyplot as plt, sys; sys.path.insert(0, "../scripts"); from md_stats import block_se
d = np.loadtxt("../data/T_series_week03.csv", delimiter=",", skiprows=1); T = d[:, 5]
def show(teq=100, b=50):
    x = T[teq:]; se1 = block_se(x, 1); seb = block_se(x, b); neff = len(x)*(se1/seb)**2
    print(f"t_eq {teq} · b {b}: mean {x.mean():.1f} K · naive SE {se1:.2f} · block SE {seb:.2f} · N_eff {neff:.0f}/{len(x)}")
    bs = [1, 2, 3, 5, 7, 10, 15, 20, 30, 50, 70, 100, 150, 200]
    fig, ax = plt.subplots(1, 2, figsize=(11, 3.4)); ax[0].plot(T, lw=1); ax[0].axvspan(0, teq, color="0.85"); ax[0].axhline(x.mean(), color="r", ls="--")
    ax[1].semilogx(bs, [block_se(x, k) for k in bs], "o-"); ax[1].plot([b], [seb], "ro", ms=10); ax[1].set_xlabel("block size"); ax[1].set_ylabel("SE (K)"); plt.show()
show(100, 50)
show(0, 1)      # 실패 상태: 초기 240 K 구간 포함 + 순진한 SE → 오차 과소
```
