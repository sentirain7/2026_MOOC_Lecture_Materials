```python
# 기본 = 내 컴퓨터 mdai 환경 · Claude Code 가 week03 폴더에서 실행 · 아래 google.colab 분기는 Colab(선택)에서만 동작
import os, sys
if "google.colab" in sys.modules and not os.path.exists("scripts/lj_verlet.py"):
    !git clone <저장소 URL> repo && %cd repo/week03
import numpy as np, matplotlib.pyplot as plt
sys.path.insert(0, "scripts")
from lj_verlet import *   # EPS SIG M CONV KB lj_u lj_f pair_step bond_sweep lj_fluid OUT
print("ε", EPS, "σ", SIG, "m", M)
```

```python
def run(dt=1.0, x0=0.10, tmax=300.0, K=478.4, r0=1.387, mu=M/2):
    omega = np.sqrt(2*K*CONV/mu); x, v = x0, 0.0; a = -2*K*x*CONV/mu; E0 = K*x*x
    ts, xs, es = [0.0], [x], [1.0]
    for i in range(1, int(tmax/dt)+1):
        v += 0.5*a*dt; x += v*dt; a = -2*K*x*CONV/mu; v += 0.5*a*dt
        ts.append(i*dt); xs.append(x); es.append((0.5*mu*v*v/CONV + K*x*x)/E0)
        if abs(x) > 5: break
    return np.array(ts), np.array(xs), np.array(es), 2*np.pi/omega
for dt in (1, 2, 4, 8):
    ts, xs, es, T = run(dt=dt)
    print(f"Δt {dt} fs · 주기 {T:.1f} fs · max|ΔE|/E₀ {'발산' if abs(xs).max()>5 else f'{abs(es-1).max()*100:.2f} %'}")
```

```python
# 그림: 신장 x(t) 와 E/E₀
fig, ax = plt.subplots(2, 1, figsize=(9, 5), sharex=True)
for dt in (1, 4, 8):
    ts, xs, es, T = run(dt=dt); ax[0].plot(ts, xs, label=f"dt {dt}"); ax[1].plot(ts, es)
ax[0].set_ylim(-0.3, 0.3); ax[1].set_ylim(0, 2); ax[0].legend(); ax[0].set_ylabel("r - r0 (A)"); ax[1].set_ylabel("E / E0"); ax[1].set_xlabel("t (fs)")
fig.tight_layout(); plt.show()
```
