```python
T = load_types()
def show(i="ca", j="hc", rule="lb"):
    m = mix(*T[i], *T[j]); r = np.linspace(0.8*min(T[i][1],T[j][1]), 8, 400)
    for k in ("lb","geo","wh"): print(k, "σ %.3f  ε %.4g  r_min %.3f" % (m[k][0], m[k][1], 2**(1/6)*m[k][0]))
    fig, ax = plt.subplots(figsize=(7,3.5))
    for k, c in [("lb","tab:orange"),("geo","tab:blue"),("wh","tab:green")]:
        s, e = m[k]; ax.plot(r, 4*e*((s/r)**12-(s/r)**6), c, lw=3 if k==rule else 1.2, label=k)
    emax = max(v[1] for v in m.values()); ax.set_ylim(-1.2*emax, emax); ax.axhline(0,color="gray",lw=.8); ax.legend(); ax.set_title(f"{i} - {j}"); plt.show()
show("ca", "hc")
show("ca", "Si_clay")     # 실패 상태: ε ≈ 4e-4 → 인력 소실
```
