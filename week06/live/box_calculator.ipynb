원본 04주차 #20 의 squalane 100 + DOCHN 100 스크립트(M 82,950 g/mol)의 구간별 셀 밀도와, 분자 수 · 분자량 · 초기 밀도에서 상자 한 변을 역산한다.

```python
# 셀 밀도 ρ = M / (0.60221 · L³)  (덱 12 · 14장 · 원본 04#20 노트)
M = 100*422.8 + 100*406.7
for name, L in [('④ 최소화', 62), ('⑤ NPT 압축', 55), ('⑥ 생산', 54)]:
    print(f'{name}: L {L} Å → ρ {M/(0.60221*L**3):.2f} g/cm³')
```

```python
# 상자 한 변 L = (N M / (ρ N_A))^(1/3) · 1 cm³ = 1e24 Å³  (덱 22장)
NA = 6.02214076e23
def L_box(n, M, rho): return (n*M/(rho*NA)*1e24)**(1/3)
print('톨루엔 50 · ρ0 0.2 →', round(L_box(50, 92.141, 0.2), 2), 'Å  (build_report.json 33.69)')
print('바인더 M̄ 480 · 80 Å · ρ 1.0 → N ≈', round(1.0*80**3/1e24*NA/480), '분자 (원본 04#34: 약 640)')
print('같은 640분자를 ρ0 0.2 로 →', round(L_box(640, 480, 0.2), 1), 'Å  (원본: 137 Å)')
print('컷오프 12 Å 조건 L/2 > 12:', L_box(50, 92.141, 0.2)/2 > 12)
```
