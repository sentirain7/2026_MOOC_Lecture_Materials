```python
import os, sys, subprocess
# 기본 = 로컬 mdai · Claude Code(RDKit · Packmol 포함) · 아래 분기는 (선택) Colab 에서만
if 'google.colab' in sys.modules:
    subprocess.run(['pip','install','-q','rdkit'], check=False); os.chdir('week11') if os.path.exists('week11') else None
SARA = '15,45,25,15'   # ← 내 측정값으로 바꾼다 (S,A,R,As wt%)
PACKMOL = 'packmol'    # 없으면 --no-pack
r = subprocess.run([sys.executable,'scripts/composition_to_counts.py','--sara',SARA,'--atoms','3000','--tag','mine','--packmol',PACKMOL], capture_output=True, text=True); print(r.stdout[-1500:], r.stderr[-800:])
```

```python
import json
rep = json.load(open('out/composition_report_mine.json'))
print('분자', rep['n_molecules'], '원자', rep['n_atoms'], 'L', rep['L_A'], 'Å')
print('wt% 오차', rep['wt_error_pct']); print('H/C', rep['HC'], 'S/C', rep['SC'])
for k,v in rep['reference_tables'].items(): print(k, 'H/C', v['HC'], 'S/C', v['SC'], '질량 분율', v['wt_from_MW'])
print(rep.get('packmol'))
```
