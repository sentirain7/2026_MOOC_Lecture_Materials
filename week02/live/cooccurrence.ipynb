# W02 라이브 · 주제 동시출현 네트워크

슬라이드 20 과 **같은 계산**. `md_construction_literature_2021_2026.csv`(267편) 의 `topics` 열(`;` 구분)에서 주제 13개의 편수와, 두 주제를 함께 가진 논문 수(동시출현)를 센다. 임계값 이상인 쌍만 선으로 그린다.

```python
import itertools, collections
import pandas as pd
df = pd.read_csv('data/md_construction_literature_2021_2026.csv')
tags = collections.Counter(); pairs = collections.Counter()
for t in df['topics'].dropna():
    ts = sorted(set(x.strip() for x in t.split(';') if x.strip()))
    tags.update(ts)
    for a, b in itertools.combinations(ts, 2):
        pairs[(a, b)] += 1
print(len(df), '편 ·', len(tags), '주제 · 공유 1편 이상 쌍', len(pairs), '/ 가능', len(tags) * (len(tags) - 1) // 2)
for k, v in tags.most_common(): print(f'{v:4d}  {k}')
```

```python
TH = 20   # 임계값 — 슬라이더 대신 숫자를 바꿔 본다
keep = {k: v for k, v in pairs.items() if v >= TH}
deg = collections.Counter()
for (a, b) in keep: deg[a] += 1; deg[b] += 1
print('표시 연결', len(keep), '쌍 · 최강', pairs.most_common(2))
print('고립 주제:', [t for t in tags if deg[t] == 0])
```

```python
# 그림 — 원 위에 주제를 놓고 임계값 이상 쌍을 선으로
import numpy as np, matplotlib.pyplot as plt
names = [k for k, _ in tags.most_common()]
ang = {n: -np.pi/2 + i*2*np.pi/len(names) for i, n in enumerate(names)}
pos = {n: (np.cos(a), np.sin(a)) for n, a in ang.items()}
fig, ax = plt.subplots(figsize=(8, 8))
for (a, b), v in keep.items():
    ax.plot([pos[a][0], pos[b][0]], [pos[a][1], pos[b][1]], color='#156082', alpha=.5, lw=1 + 6*v/101)
for n in names:
    ax.scatter(*pos[n], s=60 + 500*tags[n]/191, color='#FFFDFA', edgecolor='#156082' if deg[n] else '#C9C1B6', zorder=3)
    ax.annotate(f'{n} {tags[n]}', pos[n], fontsize=8, ha='center', va='bottom', xytext=(0, 10), textcoords='offset points')
ax.set_aspect('equal'); ax.axis('off'); ax.set_title(f'co-occurrence (threshold ≥ {TH}) · {len(keep)} / {len(pairs)} pairs')
plt.show()
```

### 해 보기
TH 를 5 · 40 · 80 으로 바꿔 가장 늦게까지 남는 쌍과 먼저 고립되는 주제를 적어 본다.
