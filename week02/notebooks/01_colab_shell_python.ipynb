```python
# 2) Python 5요소 — 변수 · 리스트 · 반복 · 함수 · import
import numpy as np          # ⑤ import
dt = 1.0                    # ① 변수: 시간 간격 (fs)
temps = [253, 273, 293]     # ② 리스트: 온도 (K)

def kinetic(m, v):          # ④ 함수: 운동 에너지 0.5 m v^2
    return 0.5 * m * v**2

for T in temps:             # ③ 반복
    print(T, 'K  →', round(kinetic(1.0, np.sqrt(T)), 1))
```

```python
# 3) CSV 읽기 — pandas 표. 267행 · 열 이름 확인
import pandas as pd
df = pd.read_csv('data/md_corpus_267.csv')
print(len(df), '행 ·', list(df.columns))
df[['year', 'title', 'venue']].head(5)
```

```python
# 4) 연도별 편수 그림 — matplotlib 동작 확인
import matplotlib.pyplot as plt
counts = df['year'].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(counts.index.astype(str), counts.values, color='#F99D1C', edgecolor='#E37F1A')
ax.set_xlabel('year'); ax.set_ylabel('papers'); ax.set_title('MD corpus 267 by year')
ax.spines[['top', 'right']].set_visible(False); plt.show()
```
