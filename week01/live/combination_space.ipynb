# W01 라이브 · 조합 공간 계산기

슬라이드 10 의 조작형 시뮬레이터와 **같은 계산**을 코드로 한다.

조합 수 = (함량 단계)^(분자 종류 수) × 온도 조건 수. 연간 실험 가능 수는 100 개로 둔다(예시 수치).

**왜** — 조합은 곱으로 늘고 실험은 더하기로 는다. 그 간격(자릿수)을 숫자로 확인한다.

```python
import math

def combinations(n_mol: int, n_level: int, n_temp: int) -> int:
    """후보 조합 수. 분자 종류마다 함량 단계를 독립으로 고른다고 가정한 상한이다."""
    return n_level ** n_mol * n_temp

EXPERIMENTS_PER_YEAR = 100   # 예시 수치 (덱 슬라이드 9·10 과 같다)

for n_mol, n_level, n_temp in [(3, 2, 4), (6, 3, 4), (13, 5, 4), (13, 10, 12)]:
    n = combinations(n_mol, n_level, n_temp)
    gap = math.log10(n / EXPERIMENTS_PER_YEAR)
    print(f'종류 {n_mol:2d} · 단계 {n_level:2d} · 온도 {n_temp:2d} → 조합 {n:.3g} · 실험 대비 {gap:5.1f} 자릿수')
```

```python
# 로그 눈금 그림 — 슬라이드와 같은 모양
import numpy as np, matplotlib.pyplot as plt
levels = np.arange(1, 11)
fig, ax = plt.subplots(figsize=(8, 4.5))
for n_mol in (3, 6, 13):
    ax.plot(levels, [combinations(n_mol, l, 4) for l in levels], marker='o', label=f'{n_mol} molecules')
ax.axhline(EXPERIMENTS_PER_YEAR, color='#2E8B57', ls='--', label='experiments / year (example)')
ax.set_yscale('log'); ax.set_xlabel('levels per molecule'); ax.set_ylabel('candidate combinations')
ax.legend(); ax.spines[['top', 'right']].set_visible(False)
plt.show()
```
