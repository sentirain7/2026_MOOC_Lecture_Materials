```python
# 1) 파이썬과 패키지 버전 — '무엇이 있는지' 를 먼저 본다
import sys, platform, json
print('python', platform.python_version())
report = {'python': platform.python_version(), 'packages': {}, 'data': {}, 'ok': False}
for name in ('numpy', 'pandas', 'matplotlib'):
    try:
        mod = __import__(name)
        report['packages'][name] = mod.__version__
        print(f'{name:12s} {mod.__version__}')
    except ImportError:
        report['packages'][name] = None
        print(f'{name:12s} 없음  ← Colab 이 아니면 pip install {name}')
```

```python
# 2) 데이터 받기 — 기본(로컬 · Claude Code)은 저장소의 data/ 를 읽는다 · Colab(선택)이면 GitHub 에서 바로 읽는다
import pandas as pd, pathlib
BASE = pathlib.Path('.') if pathlib.Path('data').exists() else pathlib.Path('..')   # week01 폴더 기준(run_nb.py · Claude Code) · notebooks/ 에서 열면 한 단계 위
local = BASE / 'data/asphalt_ml_101.csv'
URL = 'https://raw.githubusercontent.com/sentirain7/2026_MOOC_Lecture_Materials/main/week01/data/asphalt_ml_101.csv'  # 저장소 밖(Colab 등)에서 실행할 때만 사용
src = local if local.exists() else URL
df = pd.read_csv(src)
report['data'] = {'file': 'asphalt_ml_101.csv', 'rows': int(len(df)), 'columns': list(df.columns), 'years': [int(df.year.min()), int(df.year.max())]}
print(len(df), '행 ·', list(df.columns))
df.head(3)
```

```python
# 3) 그림 한 장 — 연도별 논문 수. 그림이 보이면 matplotlib 이 동작하는 것
import matplotlib.pyplot as plt
counts = df['year'].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.bar(counts.index.astype(str), counts.values, color='#F99D1C', edgecolor='#E37F1A')
ax.set_xlabel('year'); ax.set_ylabel('papers'); ax.set_title('Asphalt x ML papers (n=101) by year')
ax.spines[['top', 'right']].set_visible(False)
(BASE / 'out').mkdir(exist_ok=True)
fig.savefig(BASE / 'out/papers_by_year.png', dpi=120, bbox_inches='tight')
plt.show()
```

```python
# 4) 보고서 저장 — 이 파일 하나를 제출한다. ok 가 True 여야 한다
report['ok'] = all(report['packages'].values()) and report['data']['rows'] == 101
(BASE / 'out').mkdir(exist_ok=True)
(BASE / 'out/setup_report.json').write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding='utf-8')
print(json.dumps(report, ensure_ascii=False, indent=2))
print('\n완료:', '환경 준비됨 — out/setup_report.json 을 제출한다' if report['ok'] else '빠진 항목이 있다 — packages 에서 None 인 것을 설치한다')
```
